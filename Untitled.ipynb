```python
df = insurance_dataset
df["sex"] = pd.factorize(df["sex"])[0]
df["region"] = pd.factorize(df["region"])[0]
df["smoker"] = pd.factorize(df["smoker"])[0]
insurance_dataset = df
insurance_dataset.head()
```

```python
X = insurance_dataset.drop('expenses', axis=1)
Y = insurance_dataset['expenses']
```

```python
print(X)
print(Y)
```

```python
X_train, X_test, Y_train, Y_test= train_test_split(X,Y, test_size=0.2, random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape, Y_train.shape,Y_test.shape)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
prediction_on_train_data=model.predict(X_train)
```

```python
r2_train=metrics.r2_score(Y_train, prediction_on_train_data)
print(r2_train)
```

```python
prediction_on_test_data=model.predict(X_test)
r2_test=metrics.r2_score(Y_test, prediction_on_test_data)
print(r2_test)
```
